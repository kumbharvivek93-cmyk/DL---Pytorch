MAKING A DATA BY LINEAR REGRESSION

```python
weight=0.67
bias=0.45
start=0
end=2
step=0.02
x=torch.arange(start,end,step)
x=x.unsqueeze(dim=1)
y=bias + x*weight
```

```python
from torch.utils.data import DataLoader,TensorDataset
```

```python
data=TensorDataset(x,y)
```

```python
data=DataLoader(
    data,
    batch_size=20,
    shuffle=True
)
```
