```python
train_dataloader=DataLoader(
    shuffle=True,dataset=train_data,batch_size=64
)

test_dataloader=DataLoader(
    shuffle=False,dataset=test_data,batch_size=64
)
```

```python
class FashionMNISTmodel(nn.Module):
    def __init__(self):
        super().__init__()

        self.layer_stack=nn.Sequential(
            nn.Flatten(),
            nn.Linear(784,128),
            nn.ReLU(),
            nn.Linear(128,128),
            nn.ReLU(),
            nn.Linear(128,10)
        )
    def forward(self,x):
        return self.layer_stack(x)
```

```python
model=FashionMNISTmodel()
model
```

```python
import tqdm.auto as tqdm
```

```python
loss_function=nn.CrossEntropyLoss()
optimizer=torch.optim.SGD(lr=0.1,params=model.parameters())
```

```python

epochs = 4

for epoch in tqdm.tqdm(range(epochs)):

    # Training
    model.train()
    train_loss = 0.0

    for batch, (X, y) in enumerate(train_dataloader):

        y_train_pred = model(X)
        loss = loss_function(y_train_pred, y)

        train_loss += loss.item()

        optimizer.zero_grad()
        loss.backward()
        optimizer.step()

    train_loss /= len(train_dataloader)

    # Testing
    model.eval()
    test_loss = 0.0

    with torch.inference_mode():
        for X_test, y_test in test_dataloader:

            y_test_pred = model(X_test)
            loss_test = loss_function(y_test_pred, y_test)

            test_loss += loss_test.item()

    test_loss /= len(test_dataloader)

    if epoch % 1 == 0:
        print(
            f"Epoch: {epoch} | "
            f"Train Loss: {train_loss:.5f} | "
            f"Test Loss: {test_loss:.5f}"
        )

```
