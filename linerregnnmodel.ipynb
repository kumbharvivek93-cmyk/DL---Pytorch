```python
import torch
import torch.nn as nn
from sklearn.model_selection import train_test_split
```

```python
start=0
stop=10
step=0.01
weight=0.88
bias=0.44
```

```python
x=torch.arange(start,stop,step)
x=x.unsqueeze(dim=1)
y=weight*x+bias
```

```python
len(x)
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2)
```

```python
class LinRegModel(nn.Module):
    def __init__(self):
        super().__init__()

        self.weight=nn.Parameter(torch.randn(1,requires_grad=True,dtype=torch.float))
        self.bias=nn.Parameter(torch.randn(1,requires_grad=True,dtype=torch.float))

    def forward(self,x:torch.tensor) -> torch.tensor:
        return self.weight*x+self.bias
```

```python
torch.manual_seed(42)
model=LinRegModel()
```

```python
model.state_dict()
```

```python
loss_fun=nn.L1Loss()
optimizer=torch.optim.SGD(lr=0.001,params=model.parameters())
```

```python

epoch=1
for epoch in range (epoch):
    model.train()
    y_train_pred=model(x_train)
    loss=loss_fun(y_train_pred,y_train)
    optimizer.zero_grad()
    loss.backward()
    optimizer.step()
    model.eval()
    with torch.inference_mode():
        y_test_pred=model(x_test)
        y_test_loss=loss_fun(y_test_pred,y_test)
    if epoch % 10 ==0:
        print(f'epoch -> {epoch} , train loss -> {loss} , test loss ->{y_test_loss} ')

    
```
