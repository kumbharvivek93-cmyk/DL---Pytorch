```python
class MNISTFasionmodel(nn.Module):
    def __init__(self):
        super().__init__()

        self.layer_stack=nn.Sequential(
            nn.Flatten(),
            nn.Linear(784,100),
            nn.ReLU(),
            nn.Linear(100,100),
            nn.Softmax(dim=1),
            nn.Linear(100,10)            
        )
    def forward(self,x):
        return self.layer_stack(x)
```
